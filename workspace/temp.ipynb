```python
import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(X)

for epoch in range(epochs):

    # Forward pass
    y_pred = w * X + b

    # Loss
    mse = np.mean((y - y_pred) ** 2)

    # Gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Parameter update
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("w =", w)
print("b =", b)
print("MSE =", mse)
```
